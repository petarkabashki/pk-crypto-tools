```python
import numpy as np
import matplotlib as mpl
from matplotlib import pyplot as plt
```

```python
f = 0.025
n = 10_000
pw = 0.02
r = np.cumprod(1 + f*(np.sign(np.random.rand(n) - 0.5 + pw)))
# r
plt.plot(r)
```

```python
n = 1000
pw = 0.05
lsp = np.linspace(0, 4*pw, 40)
g = (np.sign(np.random.rand(n) - 0.5 + pw))
rs = [np.prod(1+f*g) for f in lsp]

# r
plt.semilogy(lsp, rs, '-o')
plt.axvline(2*pw)
```

```python
n = 10000
pw = 0.02
lsp = np.linspace(0, 6*pw, 40)
gs = [ (np.sign(np.random.rand(n) - 0.5 + pw)) for i in range(100)]
rs = [
    [np.prod(1+f*g) for f in lsp]
    for g in gs
]

# np.average(rs)
# r
plt.semilogy(lsp, np.array(rs).mean(axis=0), '-o')
# plt.axvline(2*pw)
```
